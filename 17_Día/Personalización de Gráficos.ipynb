Así como has visto que existe una gran cantidad de gráficos para crear con Plotly, creo que también has podido ver que al mismo tiempo existe una **gran variedad de personalizaciones de formato** para cada uno de esos gráficos.

Esto no significa, que no te pueda explicar cómo funciona la personalización de gráficos en plotly, algunos ejemplos generales, y algunas recomendaciones a tener en cuenta. Vamos a eso.

Comencemos con un gráfico muy simple al que le iremos añadiendo personalización. A continuación puedes ver que tenemos un gráfico que contiene **dos trazados**: uno que muestra datos en **barras** y otro que muestra datos en **líneas**. Hemos agregado unos elementos de **layout** muy básicos, que son los títulos del gráfico y de los ejes, y por el momento todos los aspectos estéticos del gráfico son los que trae plotly por defecto.


```python
import plotly.graph_objects as go
```

```python
fig = go.Figure()

fig.add_trace(go.Bar(x=[1, 2, 3, 4, 5],
                 y=[17, 10, 14, 15, 18],
                 name='barras'))

fig.add_trace(go.Scatter(x=[1, 2, 3, 4, 5],
                      y=[16, 12, 11, 18, 19],
                      mode='lines',
                      name='linea'))

layout = go.Layout(title='Gráfico Personalizado',
                   xaxis_title='Eje X',
                   yaxis_title='Eje Y')

fig.update_layout(layout)
```

Ahora comencemos a agregar algunas modificaciones de format a modo de ejemplo, para que conozcas las diferentes categorías en las que podemos hacer modificaciones:
1. En los elementos del gráfico
2. En los títulos
3. En los ejes
4. En la leyenda
5. En el fondo
6. En las anotaciones
7. En la información contextual

Veamos algunos ejemplos de modificaciones en estos niveles por separado.


### 1. Modificar formato de elementos gráficos del trazado

```python
# Color y ancho de línea
fig.update_traces(
    selector=dict(name='linea'),
    line=dict(color='firebrick', width=4))

# Color de la barra
fig.update_traces(
    selector=dict(name='barras'),
    marker=dict(color='lightgreen'))

# Ancho del espacio entre las barras
fig.update_layout(bargap=0.4)
```

### 2. 	Modificar el formato de títulos

```python
# Titulo principal
fig.update_layout(
    title=dict(
        text='Este es mi gráfico<br><sup>Subtítulo del Gráfico</sup>',
        font=dict(size=24, color='purple', family='Times New Roman'),
        x=0.5,
        xanchor='center'))

# Títulos de ejes
fig.update_layout(
    xaxis_title=dict(font=dict(size=18, color='blue', family='Courier New')),
    yaxis_title=dict(font=dict(size=18, color='green', family='Courier New')))
```

### 3. odificar formato de los ejes

```python
fig.update_layout(
    xaxis=dict(
        showgrid=True,
        gridcolor='lightgray',
        zeroline=True,
        zerolinecolor='gray',
        showline=True,
        linecolor='black',
        linewidth=2),
    yaxis=dict(showgrid=True,
        gridcolor='lightgray',
        zeroline=True,
        zerolinecolor='gray',
        showline=True,
        linecolor='black',
        linewidth=2))
```

### 4. Modificar parámetros de la Leyenda

```python
fig.update_layout(
    legend=dict(
        x=0.3,
        y=0.9,
        bgcolor='rgba(255, 255, 255, 0.5)',
        bordercolor='black',
        borderwidth=2,
        font=dict(size=14, color='black', family='Arial')))
```

### 5. Modificar Fondos

```python
fig.update_layout(
    plot_bgcolor='rgba(123, 65, 789, 0.5)',
    paper_bgcolor='lightblue')
```

### 6. Modificar Anotaciones

```python
# Anotación 1
fig.add_annotation(
    x=2,
    y=12,
    text="Punto Clave",
    showarrow=True,
    arrowhead=5,
    ax=20,
    ay=-30)

# Anotación 2
fig.add_annotation(
    x=3,
    y=14,
    text="Otro Punto Clave",
    showarrow=True,
    arrowhead=2,
    ax=-50,
    ay=-50,
    font=dict(
        family="Courier New, monospace",
        size=16,
        color="#ffffff"),
    align="center",
    arrowcolor="#45e6aa",
    arrowsize=1,
    arrowwidth=2,
    bordercolor="#45e6aa",
    borderwidth=2,
    borderpad=4,
    bgcolor="#666aef",
    opacity=0.5)
```

### 7. Modificar Información contextual

```python
# ejemplo 1
fig.update_traces(
    selector=dict(name='linea'),
    text=['A', 'B', 'C', 'D', 'E'],
    hoverinfo='text')
```

```python
# ejemplo 2
fig.update_traces(
    selector=dict(name='linea'),
    text=['A', 'B', 'C', 'D', 'E'],
    hovertemplate='X: %{x}<br>Y: %{y}<br>Texto: %{text}<extra>Info</extra>')
```

Estas han sido todas las áreas de los gráficos que se pueden modificar ya sea desde `update_traces()` o desde `update_layout()`.

Si bien vuelvo a mencionar que las opciones son infinitas, es bueno que puedas identificar que los cambios se pueden aplicar al nivel de los elementos del gráfico, de los títulos, de los ejes, de la leyenda, del fondo, de las anotaciones, y de la información contextual.
