### 1. Gráfico de Líneas
Comenzamos igual que en la lección anterior, creando una figura:

```python
import plotly.graph_objects as go
```

```python
fig = go.Figure()
```

Y ahora voy a hacer algo un poco diferente, en vez de crear el trazado dentro de una **variable**, y después agregarlo a la figura con `fig.add_trace()`, directamente voy a definir el trazado dentro de `fig.add_trace()` y así me salteo un paso.

```python
fig.add_trace(go.Scatter(x=[1, 2, 3, 4, 5],
                        y=[10, 15, 13, 17, 14],
                        mode='lines',
                        name='Linea'))
```

Y ahora vamos a modificar el **layout** para establecer algunos parámetros visuales de nuestro gráfico.

También te quiero mostrar que podemos hacerlo con el abordaje de la lección pasada, donde primero definimos el layout en una **variable** y luego lo actualizamos con `update_layout()`, o directamente podemos definirlo dentor de esa función sin almacenarlo en una variable.

```python
fig.update_layout(title='Gráfico de Líneas',
                 xaxis_title='Eje X',
                 yaxis_title='Eje Y')
```

### 2. Gráfico de Barras

Si bien ya hemos creado un gráfico de barras en la lección anterior, quiero mostrarte cómo puedes personalizar su formato.

Para hacer un nuevo gráfico necesito crear una figura distinta

```python
fig2 = go.Figure()
```

```python
fig2.add_trace(go.Bar(x=[1, 2, 3, 4, 5],
                    y=[5, 6, 7, 8, 9],
                    marker=dict(color='rgba(0, 128, 255, .6)',
                               line=dict(color='rgb(0, 0, 0)',
                                        width=1.5))))
```

```python
fig2.update_layout(title='Gráfico de Barras',
                 xaxis_title='Eje X',
                 yaxis_title='Eje Y')
```

### 3. Gráfico de Área

```python
fig3 = go.Figure()
```

El gráfico de área se define casi igual que un gráfico de líneas, con la diferencia de que le agregamos el parámetro `fill`. que significa *relleno*, que se encarga de colorear una determinada área del trazado.

En este ejemplo le he asignado el valor `'tozeroy'` que en inglés significa algo así como '*hasta el valor 0 del eje `y`*'.

Puedes probar otros valores para este parámetro (`'none'`, `'self'`, `'tozeroy'`, `'tonexty'`) de modo que puedas comprobar diferentes comportamientos. Recuerda hacerlo en figuras diferentes para que no se superpongan los resultados.

```python
fig3.add_trace(go.Scatter(x=[1, 2, 3, 4, 5],
                        y=[10, 15, 13, 17, 14],
                        mode='lines',
                        fill='tozeroy',
                        name='area'))

fig3.update_layout(title='Gráfico de Área',
                 xaxis_title='Eje X',
                 yaxis_title='Eje Y')
```

### 4. Gráfico Circular

```python
fig4 = go.Figure()
```

```python
fig4.add_trace(go.Pie(labels=['A', 'B', 'C ', 'D'],
                     values=[4500, 2500, 1050, 750]))

fig4.update_layout(title='Gráfico Circular')
```

### 5. Histograma

```python
data = go.Histogram(x=[1, 2, 2, 3, 3, 3, 4])

fig5 = go.Figure(data=data)

fig5.update_layout(title='Histograma')
```
