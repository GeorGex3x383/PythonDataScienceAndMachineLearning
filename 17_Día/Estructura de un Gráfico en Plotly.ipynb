# Estructura de un Gráfico en Plotly

En esta primera lección práctica, vamos a explorar la estructura básica de un gráfico en Plotly. 


### ¿Qué es Plotly?
Plotly, como puedes ver en su [sitio web oficial](https://plotly.com/python/), es una librería de **visualización de datos** que permite crear gráficos personalizados, pero lo mejor de todo, y lo que la hace muy especial, es que también puede crear **gráficos interactivos**.


### Ventajas de Plotly:
+ **Interactividad**: Los gráficos son interactivos, lo que nos permite desarrollar una exploración de datos mucho más dinámica.
+ **Personalización**: Alta capacidad de personalización en términos de estilos, colores, y formatos.
+ **Integración**: Se integra fácilmente con otras librerías de Python como *Pandas* y *NumPy*.
	
Como siempre comenzamos importando a esta librería

```python
import plotly.graph_objects as go
```

La estructura básica de un gráfico en Plotly consiste en tres componentes principales:
1. Figuras (Figures)
2. Trazados (Traces)
3. Layouts.


### Figuras (Figures)
La figura es el **contenedor principal** de un gráfico en Plotly. Adentro del `figure` se encuentran todos los elementos del gráfico, incluyendo, como ya veremos, los **trazados** y el **layout**.

Declaramos una figura asignándole una instancia de plotly.

```python
fig = go.Figure()
```

### Trazados (Traces)

Los trazados representan los **datos** que se visualizan en el gráfico. Hay diferentes tipos de trazados, como *líneas*, *barras*, *dispersión*, etc.

Comencemos haciendo un **trazado de dispersión** muy básico:

```python
mi_trazado = go.Scatter(x=[1, 2, 3, 4, 5],
                       y=[10, 11, 12, 13, 14],
                       mode='markers')
```

Una vez que hemos creado un trazado, lo añadimos a la **figura** usando el método `add_trace()`:

```python
fig.add_trace(mi_trazado)
```

Por ahora nada impresionante ¿verdad?. Pero esto puede mejorarse muchísimo, si agregamos parámetros a `mi_trazado`.

```python
mi_trazado = go.Scatter(x=[1, 2, 3, 4, 5],
                       y=[10, 11, 12, 13, 14],
                       mode='markers',
                       marker=dict(size=12,
                                  color='rgba(255, 0, 0, 0.8)',
                                  line=dict(width=2,
                                           color='DarkSlateGrey')))
```

Voy a volver a mostrar a `fig` para ver cómo han quedado estos cambios.

```python
fig
```

¡Vaya! ¡No ha cambiado nada!

Esto es porque la **figura** ha sido alimentada con la primera versión del **trazado**, y todavía no le hemos dicho a la **figura** que la variable **trazado** ha sido reescrita. 

Se lo podemos informar con la función `update_traces()`.

```python
fig.update_traces(mi_trazado)
```

Ahora solo nos falta ver el tercer elemento de los gráficos de plotly.


### Layouts
El layout define la **apariencia** del gráfico, esto incluye los *títulos*, los *ejes*, las *leyendas*, y todo lo que tenga que ver con la forma en que se van a ver nuestros gráficos.

```python
mi_layout = go.Layout(title='Ejemplo de gráfico con Plotly',
                     xaxis=dict(title='Eje de las X'),
                     yaxis=dict(title='Eje de las Y'))
```

Ahora que hemos creado un **layout propio**, podemos actualizar el **layout que viene por defecto** para los gráficos de plotly.

```python
fig.update_layout(mi_layout)
```

### Una figura, múltiples trazados

Como te dije antes, en esta `fig` podemos introducir **más de un trazado**. Entonces voy a intentar incorporar un *trazado de barras* a este mismo gráfico:

```python
trazado_barras = go.Bar(x=[1, 2, 3, 4, 5],
                       y=[5, 10, 8, 12, 7])
```

```python
fig.add_trace(trazado_barras)
```

Y algo muy interesante de los gráficos de plotly es que si **flotas con el mouse** sobre sus marcadores, vas a poder ver una etiqueta con los valores correspondientes, y este es un primer asomo de **interactividad** para el usuario, lo cual los hace mucho más interesantes.

Otro aspecto muy destacable de todos los gráficos de plotly son las **herramientas de selección y zoom** que se encuentran en la parte superior derecha del gráfico, que puedes explorar por ti mismo para ir descubriendo lo que hace cada una.

En esta lección, hemos aprendido qué es Plotly, y hemos explorado la estructura básica de un gráfico en Plotly que está basada en tres elementos: **figura**, **trazado** y **layout**.

Con esto, ya estamos listos para crear gráficos básicos y personalizarlos según nuestras necesidades. En la próxima lección, vamos a profundizar en cómo crear diferentes tipos de gráficos utilizando Plotly.
