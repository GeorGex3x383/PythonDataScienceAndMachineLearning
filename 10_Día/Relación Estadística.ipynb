```python
import seaborn as sns
```

```python
propinas = sns.load_dataset("tips")
propinas
```

```python
sns.relplot(data=propinas,x="total_bill",y="tip")
```

```python
sns.relplot(data=propinas,x="total_bill",y="tip",kind="line")
```

```python
sns.relplot(data=propinas,x="total_bill",y="tip",kind="line",hue="smoker")
```

```python
sns.relplot(data=propinas,x="total_bill",y="tip",kind="line",hue="smoker",style="smoker")
```

```python
sns.relplot(data=propinas,x="total_bill",y="tip",kind="line",hue="smoker",style="smoker",col="time")
```

```python
sns.relplot(data=propinas,x="total_bill",y="tip",kind="line",hue="smoker",style="smoker",col="time",legend=None)
```
